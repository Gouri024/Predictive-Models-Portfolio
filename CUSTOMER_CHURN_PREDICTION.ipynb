## LOGISTIC REGRESSION

```python
#  Import Libraries
import pandas as pd 
import numpy as np 
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.linear_model import LogisticRegression     # classfication model
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.metrics import accuracy_score,f1_score #model evaluation for classfication 
import joblib # To save the model
```

```python
# Data loading
df=pd.read_csv("Churn_Modelling.xls")
df.head()
```

```python
# droping useless columns 
df.drop(columns=["RowNumber","CustomerId","Surname","Geography"],inplace=True)
```

```python
df.head()
```

```python
# handling categorical columns 
df["Gender"]=df["Gender"].map({"Female":0,"Male":1})
```

```python
df.head()
```

```python
# feature selection
x=df.drop(columns=["Exited"])
y=df["Exited"]
```

```python
# data splitting 
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# model selection - comparison
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# 3 models to define
log_model = LogisticRegression(max_iter=10000)   # logistic regression for simple accuracy
rf_model = RandomForestClassifier(random_state=42)   # Random forest for medium accuracy
xgb_model = XGBClassifier(eval_metric='logloss', random_state=42)  # for best accuracy

# Best model select - XGBoost
model = xgb_model

# model train
model.fit(x_train, y_train)
```

```python
# model evaluation 
y_pred=model.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
# traning data 
y_pred_train=model.predict(x_train)
accuracy_score(y_train,y_pred_train)
```

# WITH SCALING



```python
# model traning on scaled data 
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)

x_test_scaled=scaler.transform(x_test)      # not fit and transform on test data to avoid data leakage
```

```python
# model traning o scaled data 
model.fit(x_train_scaled,y_train)
```

```python
# model evaluation on test data 
y_pred=model.predict(x_test_scaled)
accuracy_score(y_test,y_pred)
```

# PREDICTIVE MODEL 

```python
# CreditScore	Gender	Age	Tenure	Balance	NumOfProducts	HasCrCard	IsActiveMember	EstimatedSalary 
print("Enter Customer Details for Prediction:-----")
CreditScore=int(input("Enter CreditScore: "))
Gender=int(input("Enter Gender (Female:0, Male:1): "))
Age=int(input("Enter Age: "))
Tenure=int(input("Enter Tenure: "))
Balance=float(input("Enter Balance: "))
NumOfProducts=int(input("Enter NumOfProducts: "))
HasCrCard=int(input("Enter HasCrCard: "))
IsActiveMember=int(input("Enter IsActiveMember: "))
EstimatedSalary=float(input("Enter EstimatedSalary: "))

# User input to scale 
user_data=[[CreditScore,Gender,Age,Tenure,Balance,NumOfProducts,HasCrCard,IsActiveMember,EstimatedSalary]]
user_data_scaled = scaler.transform(user_data)

pred = model.predict(user_data_scaled)

if pred[0]==1:
    print("Customer will EXIT the bank")
else:
    print("Customer will STAY")
```
