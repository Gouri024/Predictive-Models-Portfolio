## CAR_PRICE_PRIDICTION

```python
# import libraries
import pandas as pd  # data preprocessing
from sklearn.model_selection import train_test_split  # for data splitting
from sklearn.linear_model import LinearRegression  # regression model 
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder,OneHotEncoder  # encoding techinique
from sklearn.metrics import r2_score  # model evaluation
```

```python
# data loading 
df=pd.read_csv(r"C:\Users\Admin\Downloads\cardekho.csv")
df.head()
```

```python
# check null values 
df.isnull().sum()
```

```python
# fiiling the null values 
df["mileage"]=df["mileage"].fillna(df["mileage"].mean())
df["engine"]=df["engine"].fillna(df["engine"].mean())
df["seats"]=df["seats"].fillna(df["seats"].median())
```

```python
df.head()
```

```python
# handling categorical  column
label=LabelEncoder()     # initializing label model
df["encoded_fuel"]=label.fit_transform(df["fuel"])
var=dict(zip(df["fuel"],df["encoded_fuel"]))
print(var)

df["encoded_seller_type"]=label.fit_transform(df["seller_type"])
var1=dict(zip(df["seller_type"],df["encoded_seller_type"]))
print(var1)

df["encoded_transmission"]=label.fit_transform(df["transmission"])
var2=dict(zip(df["transmission"],df["encoded_transmission"]))
print(var2)

df["encoded_owner"]=label.fit_transform(df["owner"])
var3=dict(zip(df["owner"],df["encoded_owner"]))
print(var3)
```

```python
df.head()
```

```python
# drop original column
df.drop(columns=["fuel","seller_type","transmission","owner"],inplace=True)
```

```python
df.head()
```

```python
# check duplicate value
df.duplicated().sum()
```

```python
# drop duplicates values 
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# feature selection for model 
x=df.drop(columns=["selling_price"])
y=df["selling_price"]
```

```python
x
```

```python
# data splitting
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# model selection
model=LinearRegression()    # model initialize

# model train
model.fit(x_train,y_train)
```

```python
# model accuracy on test data 
y_pred=model.predict(x_test)
r2_score(y_test,y_pred)
```

```python
# model accuracy on train data 
y_pred2=model.predict(x_train)
r2_score(y_train,y_pred2)
```

## PREDICTIVE MODEL

```python
# year	km_driven	mileage	engine	seats	encoded_fuel	encoded_seller_type 	encoded_transmission	encoded_owner
year=int(input("Enter the age of the car:"))
km=int(input("Enter the km_driven:"))
mileage=int(input("Enter mileage of the car:"))
engine=int(input("Enter engine power:"))
seats=int(input("Enter number of seats:"))
fuel=int(input ("Enter the fuel type --> {'Diesel': 1, 'Petrol': 3, 'LPG': 2, 'CNG': 0}:"))
seller_type=int(input("Enter the seller type --> {'Individual': 1, 'Dealer': 0, 'Trustmark Dealer': 2}:"))
transmission=int(input("Enter transmission --> {'Manual': 1, 'Automatic': 0}:"))
owner=int(input("Enter owner type --> {'First Owner': 0, 'Second Owner': 2, 'Third Owner': 4, 'Fourth & Above Owner': 1, 'Test Drive Car': 3}:"))

pred=model.predict([[year,km,mileage,engine,seats,fuel,seller_type,transmission,owner]])
print("Selling price of the car is:",pred.round(2))
```
