```python
# Import libraries 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder,OneHotEncoder  # encoding techinique
from sklearn.tree import DecisionTreeRegressor  # decision tree 
from sklearn.ensemble import RandomForestRegressor # random forest
from sklearn.metrics import r2_score
```

```python
# Data loading 
df=pd.read_csv("House_price_dataset.csv")
df.head()
```

```python
# data preprocessing   - for data cleaning 
# to check null values 
df.isnull().sum()
```

```python
# to permamnetly remove duplicated values
df.drop_duplicates(inplace=True)
```

```python
df.head()
```

```python
# To counts the values to handling categorial data 
df["date"].value_counts()
```

```python
df["street"].value_counts()
```

```python
df["city"].value_counts()
```

```python
df["statezip"].value_counts()
```

```python
df["country"].value_counts()
```

```python
# handling categorical column - 
le1 = LabelEncoder()
le2 = LabelEncoder()
df["encoded_city"] = le1.fit_transform(df["city"])
df["encoded_statezip"] = le2.fit_transform(df["statezip"])
```

```python
# then drop encoded columns
df = df.drop(columns=['encoded_date','encoded_street','encoded_country','date','street','country'], errors='ignore')
```

```python
# To check how many columns are left
print(df.columns.tolist())
df.head()
```

```python
# feature selection for model 
x=df.drop(columns=["price"])
y=df["price"]
```

```python
# data splitting 
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# MODEL ACCURACY - All models
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

# 1. Linear Regression
model = LinearRegression()
model.fit(x_train, y_train)
print("Linear Regression TRAIN:", r2_score(y_train, model.predict(x_train)))
print("Linear Regression TEST:", r2_score(y_test, model.predict(x_test)))

print("--------------------")

# 2. Decision Tree - max_depth 12 
tree = DecisionTreeRegressor(max_depth=12, random_state=42)
tree.fit(x_train, y_train)
print("Decision Tree TRAIN:", r2_score(y_train, tree.predict(x_train)))
print("Decision Tree TEST:", r2_score(y_test, tree.predict(x_test)))

print("--------------------")

# 3. Random Forest - FINAL BEST MODEL
random = RandomForestRegressor(n_estimators=200, max_depth=12, random_state=42)
random.fit(x_train, y_train)
print("Random Forest TRAIN:", r2_score(y_train, random.predict(x_train)))
print("Random Forest TEST:", r2_score(y_test, random.predict(x_test)))
```
