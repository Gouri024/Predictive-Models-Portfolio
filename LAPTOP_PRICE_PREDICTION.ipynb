```python
# Import libraries
import pandas as pd  # data preprocessing
from sklearn.model_selection import train_test_split  # for data splitting
from sklearn.linear_model import LinearRegression  # regression model 
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder,OneHotEncoder  # encoding techinique
from sklearn.metrics import r2_score  # model evaluation
```

```python
# data loading 
df=pd.read_csv("laptop_data_cleaned.csv")
df.head()
```

```python
print(df.columns.tolist())
print(df.head(2))
```

```python
# check null values 
df.isnull().sum()
```

```python
# to permanentely remove duplicate value
df.drop_duplicates(inplace=True)
```

```python
# To check duplicates value are remove or not 
df.duplicated().sum()
```

```python
# fiiling the null values 
df["Ram"]=df["Ram"].fillna(df["Ram"].mean())
df["HDD"]=df["HDD"].fillna(df["HDD"].mean())
df["SSD"]=df["SSD"].fillna(df["SSD"].median())
```

```python
# To drop the useless column 
df.drop(columns=["Company","TypeName","Cpu_brand"], inplace=True)
```

```python
# handling categorical  column
le1 = LabelEncoder()
le2 = LabelEncoder()
le3 = LabelEncoder()
le4 = LabelEncoder()
le5 = LabelEncoder()

df["encoded_Company"] = le1.fit_transform(df["Company"])
df["encoded_TypeName"] = le2.fit_transform(df["TypeName"])
df["encoded_Cpu_brand"] = le3.fit_transform(df["Cpu_brand"])
df["encoded_Gpu_brand"] = le4.fit_transform(df["Gpu_brand"])
df["encoded_Os"] = le5.fit_transform(df["Os"])
```

```python
df.head()
```

```python
# drop original column
df.drop(columns=["Company","TypeName","Cpu_brand","Gpu_brand","Os"],inplace=True)
```

```python
df.head()
```

```python
# feature selection for model 
x=df.drop(columns=["Price"])
y=df["Price"]
```

```python
# data splitting
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# model selection  - using Random forest get good accuracy
model = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
model.fit(x_train, y_train)
```

```python
# to check Acurracy both train and test data 
y_pred = model.predict(x_test)
y_pred2 = model.predict(x_train)

print("Test R2:", r2_score(y_test, y_pred))
print("Train R2:", r2_score(y_train, y_pred2))
```
