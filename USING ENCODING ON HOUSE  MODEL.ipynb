```python
# import libraries 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder,OneHotEncoder  # encoding techinique
from sklearn.tree import DecisionTreeRegressor  # decision tree 
from sklearn.ensemble import RandomForestRegressor # random forest
from sklearn.metrics import r2_score
```

```python
df=pd.read_csv(r"C:\Users\Admin\Downloads\House_price_dataset.csv")
df.head()
```

```python
# data preprocessing
# to check null values 
df.isnull().sum()
```

```python
# to permamnetly remove duplicated values
df.drop_duplicates(inplace=True)
```

```python
df.head()
```

```python
df["date"].value_counts()
```

```python
df["street"].value_counts()
```

```python
df["city"].value_counts()
```

```python
df["statezip"].value_counts()
```

```python
df["country"].value_counts()
```

```python
# handling categorical  column
label=LabelEncoder()     # initializing label model
df["encoded_date"]=label.fit_transform(df["date"])


df["encoded_street"]=label.fit_transform(df["street"])

df["encoded_city"]=label.fit_transform(df["city"])

df["encoded_statezip"]=label.fit_transform(df["statezip"])

df["encoded_country"]=label.fit_transform(df["country"])
```

```python
df.head()
```

```python
# drop original column
df.drop(columns=["date","street","city","statezip","country"],inplace=True)
```

```python
df.head()
```

```python
# feature selection for model 
x=df.drop(columns=["price"])
y=df["price"]
x
```

```python
# data splitting 
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# model selection
model=LinearRegression()    # model initialize

# model train
model.fit(x_train,y_train)
```

```python
# model testing 
y_pred=model.predict(x_test)
y_pred
```

```python
# model accuracy on test data 
y_pred=model.predict(x_test)
r2_score(y_test,y_pred)
```

```python
# model accuracy on train data 
y_pred2=model.predict(x_train)
r2_score(y_train,y_pred2)
```

```python
# model selection
tree=DecisionTreeRegressor(max_depth=7,random_state=42)  # decision initialize
random=RandomForestRegressor(n_estimators=100,max_depth=7,random_state=42) # random forest
```

```python
# model traning using decision tree 
tree.fit(x_train,y_train)
```

```python
# accuracy on test data 
y_pred=tree.predict(x_test)
r2_score(y_test,y_pred)
```

```python
# model traing using random forest
random.fit(x_train,y_train)
```

```python
# accuracy 
y_pred=random.predict(x_test)
r2_score(y_test,y_pred)
```
