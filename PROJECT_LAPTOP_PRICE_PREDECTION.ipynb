## LAPTOP_PRICE_PREDECTION

```python
# import libraries
import pandas as pd  # data preprocessing
from sklearn.model_selection import train_test_split  # for data splitting
from sklearn.linear_model import LinearRegression  # regression model 
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder,OneHotEncoder  # encoding techinique
from sklearn.metrics import r2_score  # model evaluation
```

```python
# data loading 
df=pd.read_csv(r"C:\Users\Admin\Downloads\laptop_data_cleaned.csv")
```

```python
df.head()
```

```python
# check null values 
df.isnull().sum()
```

```python
# check duplicates value 
df.duplicated().sum()
```

```python
# to permanentely remove duplicate value
df.drop_duplicates(inplace=True)
```

```python
# To check duplicates value are remove or not 
df.duplicated().sum()
```

```python
# fiiling the null values 
df["Ram"]=df["Ram"].fillna(df["Ram"].mean())
df["HDD"]=df["HDD"].fillna(df["HDD"].mean())
df["SSD"]=df["SSD"].fillna(df["SSD"].median())
```

```python
df.head()
```

```python
# handling categorical  column
label=LabelEncoder()     # initializing label model
df["encoded_Company"]=label.fit_transform(df["Company"])
var=dict(zip(df["Company"],df["encoded_Company"]))
print(var)

df["encoded_TypeName"]=label.fit_transform(df["TypeName"])
var1=dict(zip(df["TypeName"],df["encoded_TypeName"]))
print(var1)

df["encoded_Cpu_brand"]=label.fit_transform(df["Cpu_brand"])
var2=dict(zip(df["Cpu_brand"],df["encoded_Cpu_brand"]))
print(var2)

df["encoded_Gpu_brand"]=label.fit_transform(df["Gpu_brand"])
var3=dict(zip(df["Gpu_brand"],df["encoded_Gpu_brand"]))
print(var3)

df["encoded_Os"]=label.fit_transform(df["Os"])
var4=dict(zip(df["Os"],df["encoded_Os"]))
print(var4)
```

```python
df.head()
```

```python
# drop original column
df.drop(columns=["Company","TypeName","Cpu_brand","Gpu_brand","Os"],inplace=True)
```

```python
df.head()
```

```python
# feature selection for model 
x=df.drop(columns=["Price"])
y=df["Price"]
x
```

```python
# data splitting
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# model selection
model=LinearRegression()    # model initialize

# model train
model.fit(x_train,y_train)
```

```python
# model accuracy on test data 
y_pred=model.predict(x_test)
r2_score(y_test,y_pred)
```

```python
# model accuracy on train data 
y_pred2=model.predict(x_train)
r2_score(y_train,y_pred2)
```
